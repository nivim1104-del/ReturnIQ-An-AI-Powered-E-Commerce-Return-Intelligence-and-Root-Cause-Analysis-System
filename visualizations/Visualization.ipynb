# ReturnIQ: Data Visualization

##  Data Visualization

This notebook creates presentation-ready visualizations from the cleaned ReturnIQ analytical dataset.

The figures are grouped into four outputs:

1. Distribution analysis
2. Trend and delivery analysis
3. Category analysis
4. Correlation analysis

The charts are saved in the `Visualizations/` folder.

```python
%run 01_Data_Loading.ipynb
%run 02_Data_Cleaning.ipynb
%run 03_Exploratory_Analysis.ipynb
    
import os
import pandas as pd
import matplotlib.pyplot as plt

# Load the cleaned integrated dataset created during data cleaning.
df = pd.read_csv("data/cleaned/returniq_integrated_features.csv")

# Restore date columns if needed for any time-based analysis.
datetime_columns = [
    "order_purchase_timestamp",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in datetime_columns:
    if column in df.columns:
        df[column] = pd.to_datetime(df[column], errors="coerce")

order_items_complete = df

# Create the output folder for visualization images.
os.makedirs("Visualizations", exist_ok=True)

print("Dataset loaded:", order_items_complete.shape)
```

##  1. Distribution Analysis

This figure summarizes the distribution of order status and review scores.

```python

# Create a distribution figure for order status.
order_status_counts = order_items_complete["order_status"].value_counts()

plt.figure(figsize=(10, 6))
order_status_counts.plot(kind="bar")

plt.title("Order Status Distribution")
plt.xlabel("Order Status")
plt.ylabel("Number of Order Items")
plt.xticks(rotation=45)
plt.tight_layout()

plt.savefig("Visualizations/distribution_analysis.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 2. Trend and Delivery Analysis

This figure compares actual delivery time with the estimated delivery time.

```python

# Calculate average delivery and estimated delivery days by purchase month.
trend_data = (
    order_items_complete
    .dropna(subset=["order_purchase_timestamp"])
    .assign(
        purchase_month_period=lambda x:
        x["order_purchase_timestamp"].dt.to_period("M").astype(str)
    )
    .groupby("purchase_month_period")
    .agg(
        average_delivery_days=("delivery_days", "mean"),
        average_estimated_delivery_days=("estimated_delivery_days", "mean")
    )
    .reset_index()
)

plt.figure(figsize=(12, 6))
plt.plot(
    trend_data["purchase_month_period"],
    trend_data["average_delivery_days"],
    marker="o",
    label="Actual delivery"
)
plt.plot(
    trend_data["purchase_month_period"],
    trend_data["average_estimated_delivery_days"],
    marker="o",
    label="Estimated delivery"
)

plt.title("Actual vs Estimated Delivery Time")
plt.xlabel("Purchase Month")
plt.ylabel("Average Days")
plt.xticks(rotation=45)
plt.legend()
plt.tight_layout()

plt.savefig("Visualizations/trend_analysis.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 3. Category Analysis

This figure shows the most frequent product categories and helps identify high-volume categories for further investigation.

```python

# Count order items by product category.
category_counts = (
    order_items_complete["product_category_name_english"]
    .value_counts()
    .head(15)
    .sort_values()
)

plt.figure(figsize=(11, 7))
category_counts.plot(kind="barh")

plt.title("Top 15 Product Categories by Order Items")
plt.xlabel("Number of Order Items")
plt.ylabel("Product Category")
plt.tight_layout()

plt.savefig("Visualizations/category_analysis.png", dpi=300, bbox_inches="tight")
plt.show()
```

## 4. Correlation Analysis

The heatmap shows relationships among the main numerical features used during EDA. Correlation does not imply causation.

```python

# Select numerical features used in the ReturnIQ EDA.
numeric_features = [
    "price",
    "freight_value",
    "delivery_days",
    "estimated_delivery_days",
    "delivery_delay_days",
    "payment_total_value",
    "order_item_count",
    "order_total_value",
    "approval_days",
    "shipping_preparation_days",
    "transit_days",
    "seller_customer_distance_km",
    "freight_to_price_ratio",
    "product_volume_cm3",
    "review_score_mean"
]

correlation_matrix = order_items_complete[numeric_features].corr()

plt.figure(figsize=(14, 10))
plt.imshow(correlation_matrix, aspect="auto")

plt.colorbar(label="Correlation")
plt.xticks(
    range(len(numeric_features)),
    numeric_features,
    rotation=90
)
plt.yticks(
    range(len(numeric_features)),
    numeric_features
)

plt.title("Correlation Matrix of Numerical Features")
plt.tight_layout()

plt.savefig("Visualizations/correlation_analysis.png", dpi=300, bbox_inches="tight")
plt.show()
```
