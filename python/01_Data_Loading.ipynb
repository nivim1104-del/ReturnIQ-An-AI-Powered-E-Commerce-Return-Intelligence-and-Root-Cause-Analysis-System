# ReturnIQ: AI-Powered E-Commerce Return and Complaint Root-Cause Analysis System

## Problem Statement

E-commerce businesses receive many product returns and customer complaints related to wrong products,
defective or damaged products, missing products, delivery delays, non-delivery, and refund or replacement issues.

ReturnIQ is designed to analyse order, product, seller, delivery, payment, review, and complaint data
to identify patterns associated with e-commerce returns and complaints.

This notebook is responsible only for **data collection and loading**. The original datasets are not modified here.

## 1.1 Import Libraries and Load the Olist Dataset

```python
import pandas as pd

# Load the 9 CSV datasets
df_customers   = pd.read_csv('olist_customers_dataset.csv')
df_geolocation = pd.read_csv('olist_geolocation_dataset.csv')
df_order_items = pd.read_csv('olist_order_items_dataset.csv')
df_payments    = pd.read_csv('olist_order_payments_dataset.csv')
df_reviews     = pd.read_csv('olist_order_reviews_dataset.csv')
df_orders      = pd.read_csv('olist_orders_dataset.csv')
df_products    = pd.read_csv('olist_products_dataset.csv')
df_sellers     = pd.read_csv('olist_sellers_dataset.csv')
df_translation = pd.read_csv('product_category_name_translation.csv')

# Load the Excel dataset (requires 'openpyxl' library installed)
df_online_retail = pd.read_excel('online_retail_II.xlsx')
```

## 1.2 Organize the Loaded Datasets

A dictionary makes it easier to inspect multiple datasets consistently.

```python
datasets = {
    "Customers": df_customers, "Geolocation": df_geolocation, "Order Items": df_order_items,
    "Payments": df_payments, "Reviews": df_reviews, "Orders": df_orders,
    "Products": df_products, "Sellers": df_sellers, "Translation": df_translation,
    "Online Retail II": df_online_retail
}

for name, df in datasets.items():
    print(f"{name}: {df.shape[0]} rows, {df.shape[1]} columns")
```

## 1.3 Inspect Dataset Structure

```python
# Inspect the datasets
for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Column names:")
    print(df.columns.tolist())
    print()
```

## 1.4 Check Missing Values

This is an initial inspection only. Missing values are handled in the cleaning notebook.

```python
for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print(df.isnull().sum())
    print()
```

## 1.5 Check Duplicate Rows

Duplicate records are inspected before cleaning.

```python
for name, df in datasets.items():
    print("=" * 60)
    print(name)
    print("Duplicate rows:", df.duplicated().sum())
    print()
```
